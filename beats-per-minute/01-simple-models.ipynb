```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df= pd.read_csv('train.csv')
df.set_index('id')
df
```

```python
# Zakładamy, że df to Twój DataFrame
# Wybieramy cechy numeryczne bez id i targetu
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

features = df.drop(columns=['BeatsPerMinute', 'id']).columns

X = df[features]
y = df['BeatsPerMinute']

# Skalowanie cech – przydatne dla Lasso i Ridge
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Podział na zbiór treningowy i testowy (80/20)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Tworzymy model
lr = LinearRegression()
lr.fit(X_train, y_train)

# Predykcje
y_pred_lr = lr.predict(X_test)

# Ewaluacja
print("Linear Regression")
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_lr)))
print("MAE:", mean_absolute_error(y_test, y_pred_lr))
print("R2:", r2_score(y_test, y_pred_lr))
```

```python
from sklearn.linear_model import Ridge

ridge = Ridge(alpha=1.0)  # alpha – siła regularyzacji
ridge.fit(X_train, y_train)
y_pred_ridge = ridge.predict(X_test)

print("\nRidge Regression")
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_ridge)))
print("MAE:", mean_absolute_error(y_test, y_pred_ridge))
print("R2:", r2_score(y_test, y_pred_ridge))
```

```python
from sklearn.linear_model import Lasso

lasso = Lasso(alpha=0.01)  # alpha – siła regularyzacji
lasso.fit(X_train, y_train)
y_pred_lasso = lasso.predict(X_test)

print("\nLasso Regression")
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_lasso)))
print("MAE:", mean_absolute_error(y_test, y_pred_lasso))
print("R2:", r2_score(y_test, y_pred_lasso))
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# tworzymy model
mlp = MLPRegressor(
    hidden_layer_sizes=(64, 64),   # 2 warstwy po 64 neurony
    activation='relu',             # funkcja aktywacji
    solver='adam',                 # optymalizator
    max_iter=500,                 # maksymalna liczba iteracji
    random_state=42,
    verbose = True
)

mlp.fit(X_train, y_train)

# predykcje
y_pred = mlp.predict(X_test)

# ewaluacja
print("MLPRegressor")
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2:", r2_score(y_test, y_pred))
```
