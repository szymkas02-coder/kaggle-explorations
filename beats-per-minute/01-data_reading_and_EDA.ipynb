```python
# Przykład: df to Twój DataFrame
features = df.columns.drop(['BeatsPerMinute', 'id'])
```

```python
plt.figure(figsize=(15,10))
for i, col in enumerate(features):
    i = int(i)
    plt.subplot(3, 3, i+1)
    
    # Hexbin plot – pokazuje gęstość punktów
    plt.hexbin(df[col], df['BeatsPerMinute'], gridsize=50, cmap='viridis', mincnt=1)
    
    plt.xlabel(col)
    plt.ylabel('BeatsPerMinute')
    
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Zakładamy, że df jest Twoim DataFrame

# -------------------------
# 1. Tworzenie nowych cech
# -------------------------
df['Energy_x_Rhythm'] = df['Energy'] * df['RhythmScore']          # cecha interakcyjna
df['log_AudioLoudness'] = np.log1p(df['AudioLoudness'] - df['AudioLoudness'].min() + 1)  # log-transform, unikamy log(0)
df['Vocal_Content_Squared'] = df['VocalContent'] ** 2             # np. kwadrat cechy
df['log_Energy'] = np.log1p(df['Energy'] - df['Energy'].min() + 1)  # log-transform, unikamy log(0)

# -------------------------
# 2. Lista nowych cech do sprawdzenia
# -------------------------
new_features = ['Energy_x_Rhythm', 'log_AudioLoudness', 'Vocal_Content_Squared', 'log_Energy']

# -------------------------
# 3. Hexbin plot dla nowych cech vs BeatsPerMinute
# -------------------------
plt.figure(figsize=(15,5))

for i, col in enumerate(new_features):
    plt.subplot(1, len(new_features), i+1)
    plt.hexbin(df[col], df['BeatsPerMinute'], gridsize=50, cmap='viridis', mincnt=1)
    plt.xlabel(col)
    plt.ylabel('BeatsPerMinute')
    plt.colorbar(label='Liczba punktów')

plt.tight_layout()
plt.show()
```
